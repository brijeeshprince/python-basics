```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

if(age >= 21 and age <= 30):
    degree = input("Enter your degree: ")
    if(degree == "B.Tech" or degree == "BE" or degree == "BCA" or degree == "MCA"):
        percentage = int(input("Enter your percentage: "))
        if(percentage >= 70):
            python = input("Do you know python? (yes/no): ")
            java = input("Do you know java? (yes/no): ")
            if(python == "yes" or java == "yes"):
                backlogs = input("Do you have any backlogs? (yes/no): ")
                if(backlogs == "no"):
                    passport = input("Do you have a passport? (yes/no): ")
                    if(passport == "yes"):
                        print(name,"You are eligible for the job.")
                    else:
                        print("You are not eligible due to passport criteria.")
                else:
                    print("You are not eligible due to backlogs.")
            else:
                print("You are not eligible due to programming knowledge.")
        else:
            print("You are not eligible due to percentage criteria.")
    else:
        print("You are not eligible due to degree criteria.")
else:
    print("You are not eligible due to age criteria.")
```
