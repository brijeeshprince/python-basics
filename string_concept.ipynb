```python
Student_Name=input("Name: ")
Email_id=input("Email: ")
Phone_Number=input("Phone : ")
Course=input("Course: ")
print("="*8 , "STUDENT PROFILE" , "="*8,"\n")
print("Name             : " , Student_Name)
print("Email            : " , Email_id)
print("Phone            : " , Phone_Number)
print("Course           : " , Course,"\n")
print("-"*8 , "STRING ANALYSIS" , "-"*8,"\n")
print("Name length      : " , len(Student_Name))
print("Name Upper       : " , Student_Name.upper())
print("Name Lower       : " , Student_Name.lower())
print("First Name       : " , Student_Name[0])
print("Last Character   : " , Student_Name[-1])
print("\nEmail Domain     : " , Email_id.split("@")[1])
print("Course Length    : ", len(Course))
print("="*35)
```
