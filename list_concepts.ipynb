```python
marks=[80,90,70,60,50]
print("maximum mark:",max(marks))
print("minimum mark:",min(marks))
average=sum(marks)/len(marks)
print("average mark:",average)
```

```python
print(marks[0:3])
print(marks[1:2])
print(marks[:2])
print(marks[2:])
```

```python
marks.sort()
new=marks.copy()
print(marks[::-1])
new.append("apple")
print(new)
print(marks)
```
